```python
movie_map = df_movies[["movieId", "movie_name"]]
```

```python
df_movie_final = pd.merge(df, movie_map, left_on="movieId", right_on="movieId", how="inner")
```

```python
df_movie_final.head()
```

```python
popular_movies = (
    df_movie_final.groupby('movieId')
      .agg(avg_rating=('rating', 'mean'), count=('rating', 'count'))
      .reset_index()
)

# Filter strong candidates
popular_movies = popular_movies[
    (popular_movies['count'] > 50) & (popular_movies['avg_rating'] >= 3.5)
]
```

```python
popular_movies
```

```python
popular_movies = popular_movies.merge(
   df_movies[['movieId', 'movie_name', 'genres']],
    on='movieId'
)
```

```python
popular_movies
```

```python
cold_start_candidates = (
    popular_movies
    .explode('genres')
    .sort_values(['genres', 'avg_rating', 'count'], ascending=[True, False, False])
    .groupby('genres')
    .head(2)   # 2 movies per genre
)
```

```python
final_movies = cold_start_candidates.sample(30, random_state=42)
```

```python
df_data = final_movies[['movieId', 'movie_name', 'genres']]
```
