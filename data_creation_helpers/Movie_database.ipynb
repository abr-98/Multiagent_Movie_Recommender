```python
import pandas as pd
```

```python
df_imdb = pd.read_csv('/content/drive/MyDrive/data_movie/data_movie/imdb_dataset.csv')
```

```python
df_imdb.head()
```

```python
df_imdb['Genre'] = df_imdb['Genre'].apply(lambda x: x.split(','))
```

```python
df_imdb.head()
```

```python
df_imdb.drop("Meta_score", axis=1, inplace=True)
```

```python
df_imdb.dropna(inplace=True)
```

```python
df_imdb["Gross"] = df_imdb["Gross"].apply(lambda x: int(str(x).replace(",","")))
```

```python
df_imdb.head()
```

```python
df_imdb.rename(columns={"Series_Title": "Title", "IMDB_Rating": "Rating"}, inplace=True)
```

```python
df_imdb.head()
```

```python
df_imdb.rename(columns={"cast": "Cast"}, inplace=True)
```

```python
df_imdb.head()
```

```python
df_imdb["Runtime"] = df_imdb["Runtime"].apply(lambda x: int(x.split()[0]))
```

```python
df_imdb.head()
```

```python
df_tmdb = pd.read_csv('/content/drive/MyDrive/data_movie/data_movie/TMDB_final.csv')
```

```python
df_tmdb.head()
```

```python
df_tmdb=df_tmdb.drop("id", axis=1)
```

```python
df_tmdb.head()
```

```python
df_tmdb=df_tmdb.drop("movie_id", axis=1)
```

```python
df_tmdb.dropna(inplace=True)
```

```python
df_tmdb["runtime"] = df_tmdb["runtime"].apply(lambda x: int(x))
```

```python
df_tmdb.rename(columns={"title": "Title", "vote_average": "Rating", "genres": "Genre", "overview": "Overview", "cast": "Cast", "revenue": "Gross", "runtime" : "Runtime"}, inplace=True)
```

```python
df_tmdb["Released_Year"] = df_tmdb["release_date"].apply(lambda x: int(x.split("-")[0]))
```

```python
df_tmdb.drop("release_date", axis=1, inplace=True)
```

```python
df_final = pd.concat([df_imdb, df_tmdb], ignore_index=True)
```

```python
df_final.head()
```

```python
df_final.tail()
```

```python
df_final.fillna("", inplace=True)
```

```python
df_final.to_csv("/content/drive/MyDrive/data_movie/data_movie/final_database.csv", index=False)
```
